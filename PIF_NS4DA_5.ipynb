### **1.2. Descripción del problema**
---

Resumen breve del problema resuelto.

Los incendios de la cobertura vegetal en Colombia no ocurren aislados: municipios vecinos suelen tener incendios en los mismos meses, porque comparten clima, vegetación y temporada seca. Las series de tiempo y los mapas de puntos muestran *dónde* y *cuándo* hay incendios, pero no qué zonas están conectadas por su actividad ni qué municipios unen regiones de riesgo distintas.

Además, el fenómeno de El Niño reduce las lluvias en buena parte del país y se asocia con más incendios. No está claro si, además de aumentar la cantidad de incendios, El Niño cambia la **estructura** del riesgo: si lo vuelve más conectado entre regiones.

Siguiendo la idea de Ferreira et al. (2020), representamos la actividad de incendios como una red: los municipios son nodos, y dos municipios cercanos se conectan si tienen incendios el mismo mes. Sobre esa red aplicamos los métodos de Matta et al. (2026) para responder:

> **¿Qué municipios conectan el riesgo de incendio en Colombia y cómo cambia esa estructura entre El Niño y La Niña (2002–2024)?**

### **1.6. Enfoque de Ciencia de Redes implementado**
---

Describe en términos generales el enfoque de ciencia de redes implementado. Se deben mencionar el(los) tipo(s) de modelo(s) y método(s) de ciencia de redes que utilizaron.
Nota adicional: En este proyecto no es necesario realizar una evaluación estadística de los métodos y modelos de ciencia de redes implementados, ya que esto queda fuera del alcance de la clase.

La red es **no dirigida, ponderada y espacial**: las aristas solo pueden existir entre municipios a menos de 50 km. Los métodos se agrupan en seis bloques, siguiendo a Matta et al. (2026), más la extensión ENSO:

| Bloque | Métodos | Pregunta |
|---|---|---|
| Descriptivo | Grado, fuerza, clustering, componente gigante, camino promedio, distribución de grado | ¿Cómo es la red? |
| Modelos nulos | Erdős–Rényi, modelo de configuración, permutación de meses y de años | ¿Es más que azar? ¿Es un mundo pequeño? |
| Comunidades | Louvain (Blondel et al., 2008; peso `w_jaccard`, mejor de 30 semillas), modularidad Q; comparación con departamentos (ARI, NMI) | ¿Hay regiones de riesgo? |
| Hubs vs. puentes | Top 15 por grado frente a top 15 por intermediación (*betweenness*); coeficiente de participación | ¿Qué municipios unen regiones? |
| Resiliencia | Remoción adaptativa por grado, por intermediación y al azar (100 réplicas); curva de la componente gigante y su área (AUC) | ¿Dónde priorizar la prevención? |
| Difusión | SIR de tiempo discreto (paso = 1 mes) con parámetros calibrados con los datos; potencial propagador; inmunización de hubs, puentes o al azar | ¿Desde dónde se propaga el riesgo? |
| Extensión ENSO | Una red por fase; métricas por fase; prueba de permutación de las etiquetas de fase de los meses (1.000 permutaciones, en tres versiones) | ¿El Niño conecta más el riesgo? |

**Calibración del SIR.** Los parámetros se estimaron con la tabla municipio–mes, no se eligieron a mano:
- **γ** (probabilidad mensual de dejar de estar activo) = 1 − P(activo en t+1 | activo en t).
- **β** (probabilidad mensual de que un vecino activo active a un municipio inactivo), por máxima verosimilitud con el modelo P(activarse | k vecinos activos) = 1 − (1 − p₀)(1 − β)^k, donde p₀ es la activación sin vecinos activos.
- Se compara la transmisibilidad por arista T = β / (1 − (1 − β)(1 − γ)) con el umbral epidémico de la red T_c = ⟨k⟩ / (⟨k²⟩ − ⟨k⟩) (Newman, 2002; Pastor-Satorras et al., 2015).

**Prueba de permutación ENSO.** Se barajan las etiquetas de fase entre los meses (cada fase conserva su número de meses) y se recalcula la diferencia El Niño − La Niña. Tres versiones: libre; solo entre meses del mismo mes calendario, para controlar la estacionalidad; y solo entre meses del mismo año, para controlar que el reporte del IDEAM crece con los años.

Herramientas: Python con NetworkX, igraph, pandas, GeoPandas y Matplotlib.

#### 1.11.5. Difusión SIR: El Niño lleva la red al umbral de propagación

**Parámetros calibrados con los datos** (paso = 1 mes):

| | γ | Duración media (meses) | β | T por arista | T / T_c |
|---|---|---|---|---|---|
| Todos los meses | 0,65 | 1,5 | 0,020 | 0,030 | 0,93 |
| El Niño | 0,62 | 1,6 | 0,021 | 0,034 | **1,03** |
| La Niña | 0,71 | 1,4 | 0,017 | 0,023 | **0,71** |

- Un municipio activo sigue activo al mes siguiente el 35 % de las veces (γ = 0,65): un episodio dura en promedio 1,5 meses.

**Hubs y puentes sirven para objetivos distintos.** La resiliencia y el SIR parecen contradecirse, pero responden preguntas diferentes:
- Para que un brote **sea más pequeño**, conviene proteger los **hubs** del altiplano: ahí se originan y crecen los brotes.
- Para que las regiones **no queden conectadas entre sí**, conviene proteger los **puentes** del corredor Cesar–Magdalena Medio–Córdoba–Antioquia: quitarlos parte la red en regiones, aunque no achica los brotes dentro de cada región.

### **1.12. Conclusiones**
---

Presenta las conclusiones del trabajo, aborda las limitaciones de su análisis y cómo este podría mejorarse o desarrollarse a partir del trabajo realizado en la clase.

**Respuesta a la pregunta.** El riesgo de incendio en Colombia está organizado en regiones muy cohesionadas, unidas por pocos municipios puente del corredor Cesar–Magdalena Medio–Córdoba–Antioquia, que no coinciden con los municipios de más vecinos (los hubs del altiplano cundiboyacense). **El Niño conecta más el riesgo:** la red pasa de estar partida en regiones durante La Niña (componente gigante del 56 %) a ser casi una sola red nacional (92 %), y queda justo por encima del umbral en el que el riesgo puede propagarse a escala regional. La diferencia supera a todas las permutaciones aleatorias de la fase de los meses, también al controlar la estacionalidad y el aumento del reporte. El mecanismo es la simultaneidad: en El Niño se queman más del doble de municipios a la vez.

**Recomendaciones de prevención**
1. **Para reducir el tamaño de los brotes:** reforzar la prevención y la respuesta en el altiplano cundiboyacense en años de El Niño. Es donde se originan los brotes más grandes y donde vacunar tiene más efecto en la simulación.
2. **Para evitar que las regiones se conecten:** vigilar el corredor Cesar–Magdalena Medio–Córdoba–Antioquia (San Alberto, Sabana de Torres, Pueblo Nuevo, Ituango). Quitar el 5 % de estos municipios parte la red a la mitad.
3. **Coordinar por región y no solo por departamento:** las regiones de riesgo cruzan los límites departamentales.
4. **Usar el índice ONI como alerta temprana:** El Niño se anuncia con meses de anticipación, y con él la red cruza el umbral de propagación.

**Limitaciones**
- Una arista es co-ocurrencia de registros en el mismo mes, no propagación física del fuego. Gran parte de la co-ocurrencia se debe a causas comunes, como el clima compartido entre vecinos; por eso el β del SIR es una cota superior del contagio real.
- El reporte del IDEAM crece mucho con los años (2024 tiene unas 9 veces los registros de 2002) y puede depender de la capacidad de reporte de cada entidad. Se controló con la permutación por año y la sensibilidad sin 2023–2024, pero no se puede eliminar del todo.
- Sin coordenadas desde 2021, la red es municipal: un municipio grande cuenta igual que uno pequeño, y las distancias se miden entre centroides.
- La intermediación y el SIR se calcularon sin pesos; el umbral T_c es de campo medio, y en una red tan agrupada el umbral real es algo mayor.
- No se usó el área quemada, porque su unidad no está confirmada.

**Cómo mejorar o extender el trabajo**
- Usar detecciones satelitales (MODIS o VIIRS) para tener coordenadas y una serie homogénea en el tiempo, sin el sesgo de reporte.
- Construir redes dirigidas (municipio activo en t → vecino activo en t+1), como las redes cronológicas de Ferreira et al. (2020), para acercarse más a la propagación.
- Incluir el área quemada y variables de clima y cobertura vegetal en los pesos, o en un SIR con activación espontánea.
- Repetir el análisis con ventanas de 3 meses y con los umbrales de 25 y 75 km.
